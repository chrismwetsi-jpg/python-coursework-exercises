INDEXING IN PYTHON
indexing starts at 0 
negative indexing starts at -1 from end 

```python
# INDEXING IN PYTHON

# String indexing
text = "chris"
print(text[0]) 
print(text[-1]) 
print(text[-2]) 

# List indexing
fruits = ["apple", "banana", "mango", "orange"]
print(fruits[0]) 
print(fruits[-1])

# Slicing [start:stop] - stop not included
print(text[0:4]) 
print(text[2:]) 
print(text[:4]) 
print(text[::2]) 
print(text[::-1])

# List slicing
print(fruits[1:3]) # ['banana', 'mango']
print(fruits[:2]) # first two
print(fruits[::-1]) # reversed list

# 2D indexing (list inside list)
matrix = [[1,2,3], [4,5,6], [7,8,9]]
print(matrix[0]) # first row
print(matrix[0][1]) # 2 - first row, second column
print(matrix[2][2]) # 9
```

OPERATION IN PYTHON
Arthimetic,Comparison, Logical, Assignment, Membership, Identity

```python
a = 10
b = 3
#
# Arithmetic
print("a + b =", a + b)#Addition
print("a - b =", a - b)#substraction
print("a * b =", a * b)#Multiplication
print("a / b =", a / b)#Division
print("a % b =", a % b) # remainder
print("a ** b =", a ** b) # power
print("a // b =", a // b) # floor division

# Comparison
#when we are comparing numbers it  gives output in boolean
print(a > b, a < b, a == b, a!= b)

# Logical
#outputs are in boolean
print((a > 5) and (b < 5))
print((a > 15) or (b < 5))
print(not (a == 10))

# Membership
#outputs are always in boolean
fruits = ["apple", "banana"]
print("apple" in fruits)
print("mango" not in fruits)
```

DATA STRUCTURE IN PYTHON
(way to organise data) 

```python
# Stack 
stack = [33,4,5]
stack.append(10)
stack.append(20)
stack.append(30)
stack.remove(33)
print("Stack:", stack)
print("Pop:", stack.pop())
print("After pop:", stack)


# Nested data structure - real example
students = [
    {"name": "Chris", "scores": [85, 90]},
    {"name": "Bahati", "scores": [92, 88]}
]
for s in students:
    avg = sum(s["scores"])/len(s["scores"])
    print(f"{s['name']} avg: {avg}")
```

COLLECTION: list, Tuple, set. dictionnary

```python
# LIST - ordered, changeable, allows duplicates
my_list = [1, 2, 3, 4]
my_list.append(5)


print("List:", my_list)

# TUPLE - ordered, unchangeable
my_tuple = (1, 2, 3)
print("Tuple:", my_tuple, "len:", len(my_tuple))

# SET - unordered, no duplicates
my_set = {1, 2, 2, 3, 3, 4}
print("Set:", my_set)
my_set.add(5)
print("After add:", my_set)

# DICTIONARY - key:value
person = {"name": "Chris", "age": 20, "city": "Goma"}
print(person["name"])
person["age"] = 21
print(person)
print("Keys:", person.keys())
print("Values:", person.values())
```

CONTROL FLOW_ Break,Continue, pass

```python
# break
for i in range(1, 10):
    if i == 7:
        break
    print(i, end=" ")
print("\nBroke at 7")

# continue
for i in range(1, 6):
    if i == 3:
        continue
    print(f"Number {i}")

# pass - placeholder
for i in range(3):
    if i == 1:
        pass 
    print(f"Pass demo {i}")
```

LOOPS

```python
# for loop
for i in range(5):
    print(i)

fruits = ["apple", "mango", "banana"]
for f in fruits:
    print(f"I like {f}")

# while loop
count = 0
while count < 3:
    print("Hello")
    count += 1

# Nested loop
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i*j}")
```

LAMBDA FUNCTION

```python
# lambda arguments: expression

square = lambda x: x * x
print(square(5))

add = lambda a, b: a + b
print(add(3, 7))

# Useful with map and filter
nums = [1,2,3,4,5]
squared = list(map(lambda x: x**2, nums))
print(squared)

evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
```

 OBJECT ORIENTED PROGRAMMING

```python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def introduce(self):
        print(f"I am {self.name}, {self.age}, studying {self.course}")

s1 = Student("Chris", 20, "Python")
s2 = Student("Aline", 22, "Data Science")

s1.introduce()
s2.introduce()
```
